```python
import datetime
import pandas as pd
import pygmt

# ==========================================
# 1. 參數與區域設定 (夏威夷熱點火山 - 多條剖面)
# ==========================================
# 地圖範圍：涵蓋夏威夷群島 (西經 165°W ~ 152°W，北緯 17°N ~ 24°N)
region = [-165, -152, 17, 24]

# 定義多條剖面線 (使用字典儲存名稱、起點與終點)
# 這三條線皆「垂直」於夏威夷島鏈走向，分別切過不同年代的島嶼
profiles = {
    "A-B (Big Island/Active)": {"start": [-157.0, 18.0], "end": [-154.0, 21.0]},  # 年輕、活躍熱點 (夏威夷大島)
    "C-D (Oahu/Mid-age)":      {"start": [-159.0, 19.5], "end": [-156.0, 22.5]},  # 中期沉降島嶼 (歐胡島/茂宜島)
    "E-F (Kauai/Old)":         {"start": [-161.0, 21.0], "end": [-158.0, 24.0]},  # 古老受侵蝕島嶼 (考艾島)
}

corridor_width = 150  # 走廊全寬 (km)

start_time = "2000-01-01"
end_time = datetime.datetime.now().strftime("%Y-%m-%d")

# ==========================================
# 2. 自動查詢 USGS 地震資料 (超過 20,000 筆自動提高 M)
# ==========================================
min_magnitude = 5.0
max_events = 20000
df_eq = pd.DataFrame()

print("正在向 USGS 查詢地震資料...")

while True:
    url = (
        f"https://earthquake.usgs.gov/fdsnws/event/1/query?format=csv"
        f"&starttime={start_time}&endtime={end_time}"
        f"&minmagnitude={min_magnitude}"
        f"&minlongitude={region[0]}&maxlongitude={region[1]}"
        f"&minlatitude={region[2]}&maxlatitude={region[3]}"
    )
    try:
        temp_df = pd.read_csv(url)
        count = len(temp_df)
        print(f"規模 M >= {min_magnitude:.1f}：搜尋到 {count} 筆地震。")

        if count <= max_events:
            df_eq = temp_df
            break
        else:
            print(
                f"⚠️ 筆數超過 {max_events} 筆限制，調高規模至 M >= {min_magnitude + 0.2:.1f} 重試..."
            )
            min_magnitude += 0.2
    except Exception as e:
        print(f"下載失敗: {e}")
        break

# 整理地震資料 (順序：經度, 緯度, 深度)
eq_data = df_eq[["longitude", "latitude", "depth"]].dropna()

# ==========================================
# 3. 定義三段深度色階 (CPT)
# 0–70 km (紅色), 70–300 km (綠色), 300–700 km (藍色)
# ==========================================
cpt_content = """0	red	70	red
70	green	300	green
300	blue	700	blue
"""
with open("eq_depth.cpt", "w") as f:
    f.write(cpt_content)


# ==========================================
# 4. 繪製平面地圖 (region_map.png)
# ==========================================
fig_map = pygmt.Figure()

grid = pygmt.datasets.load_earth_relief(resolution="02m", region=region)
fig_map.grdimage(grid=grid, cmap="geo", frame=True)
fig_map.coast(shorelines="0.5p,black", borders="1/0.5p,gray")

# 畫地震點
fig_map.plot(
    data=eq_data, style="c0.12c", fill="+z", cmap="eq_depth.cpt", pen="0.2p,black"
)

# 迴圈畫出多條剖面線與端點標籤
label_pairs = [["A", "B"], ["C", "D"], ["E", "F"]]
for i, (name, coords) in enumerate(profiles.items()):
    pt_start = coords["start"]
    pt_end = coords["end"]

    # 畫紅線
    fig_map.plot(x=[pt_start[0], pt_end[0]], y=[pt_start[1], pt_end[1]], pen="2p,red")
    # 標示端點代號 (A-B, C-D, E-F)
    fig_map.text(
        x=[pt_start[0], pt_end[0]], y=[pt_start[1], pt_end[1]],
        text=label_pairs[i], font="14p,Helvetica-Bold,red", justify="CM"
    )

fig_map.basemap(
    map_scale="jBL+w200k+o0.5c/0.5c+f+l",
    frame=['+t"Hawaiian Hotspot Seismicity"', f'+s"Data: USGS (M>={min_magnitude:.1f}, N={len(eq_data)})"'],
)
fig_map.savefig("region_map.png")
print("✅ 已成功儲存平面圖: region_map.png")

# ==========================================
# 5. 計算並繪製多條剖面圖 (輸出多張圖片)
# ==========================================
label_pairs = [["A", "B"], ["C", "D"], ["E", "F"]]

for i, (name, coords) in enumerate(profiles.items()):
    pt_start = coords["start"]
    pt_end = coords["end"]
    label = f"{label_pairs[i][0]}-{label_pairs[i][1]}"  # 組合出 "A-B", "C-D" 等字串

    # 將地震資料投射到該剖面上
    projected_eq = pygmt.project(
        data=eq_data,
        center=pt_start,
        endpoint=pt_end,
        width=[-corridor_width/2, corridor_width/2],
        unit=True,
    )

    fig_sec = pygmt.Figure()

    # 處理投影後的資料
    sec_data = projected_eq.iloc[:, [3, 2, 2]] if len(projected_eq) > 0 else pd.DataFrame()
    profile_length = float(sec_data.iloc[:, 0].max()) if len(sec_data) > 0 else 500
    ve_factor = round((100 / profile_length) * (15 / 10), 2)  # 夏威夷地震較淺，深度軸先設100km計算VE

    # 設定座標軸 (夏威夷地震極淺，深度軸設 0~100 km 即可觀察岩漿庫與地殼活動)
    fig_sec.basemap(
        region=[0, profile_length, 0, 100],
        projection="X15c/-10c",
        frame=[
            'xaf+l"Distance along profile (km)"',
            'ya20f10+l"Depth (km)"',
            f'WSne+t"Cross Section {label} ({name.split(" ")[1]} VE={ve_factor}x)"',
        ],
    )

    # 繪製剖面點
    if len(sec_data) > 0:
        fig_sec.plot(
            data=sec_data, style="c0.15c", fill="+z", cmap="eq_depth.cpt", pen="0.2p,black"
        )

    fig_sec.text(
        x=profile_length * 0.02, y=95,
        text=f"Source: USGS | Corridor: {corridor_width}km | Events: {len(sec_data)}",
        font="10p,Helvetica,black", justify="ML"
    )

    filename = f"region_section_{label}.png"
    fig_sec.savefig(filename)
    print(f"✅ 已成功儲存剖面圖: {filename}")
```

```python
# ==========================================
# 6. 繪製 3D 立體透視圖 (region_3d.png)
# ==========================================
fig_3d = pygmt.Figure()

# 設定 3D 繪圖範圍：[西經120, 西經95, 南緯15, 北緯5, 深度-8000m, 高程2000m]
region_3d = [region[0], region[1], region[2], region[3], -8000, 2000]

# 繪製 3D 地形表面 (grdview)
fig_3d.grdview(
    grid=grid,
    region=region_3d,
    projection="M15c",
    zscale="0.001c",  # Z 軸高程縮放倍率
    perspective=[135, 30],  # [方位角 135 度, 仰角 30 度]
    surftype="s",  # 's' 表示繪製連續曲面 (Surface)
    cmap="geo",
    plane="-8000+ggray",  # 底座填滿灰色
    frame=[
        'xa5f2.5+l"Longitude"',
        'ya5f2.5+l"Latitude"',
        'za2000f1000+l"Elevation (m)"',
        'WSneZ+t"3D Perspective View of East Pacific Rise"',
    ],
)

# 在 3D 地形曲面上投射地震點
# 注意：3D 繪圖需傳入 (經度, 緯度, 高程)，地震深度需轉為負的高程值 (米)
eq_3d_data = eq_data.copy()
eq_3d_data["elevation"] = eq_3d_data["depth"] * -1000  # km 轉成 m 並取負值

fig_3d.plot3d(
    data=eq_3d_data[["longitude", "latitude", "elevation", "depth"]],
    style="s0.15c",  # 3D 立體方塊 (s) 或 球體 (c)
    fill="+z",
    cmap="eq_depth.cpt",
    perspective=[135, 30],
    pen="0.1p,black",
)

fig_3d.savefig("region_3d.png")
print("✅ 已成功儲存 3D 立體圖: region_3d.png")
```

```python
import pandas as pd
import pygmt

# ==========================================
# 1. 地圖區域與剖面線設定 (夏威夷群島區域)
# 經度使用 195~210 (即西經 165W~150W 的正值表示，避免跨越 180 度線問題)
# ==========================================

# 標示夏威夷熱點 (Hotspot Location)
hotspot_lon, hotspot_lat = 204.7, 19.3  # 夏威夷大島南側 Lōʻihi 海底火山附近

# 垂直於島鏈走向的 3 條剖面線
profiles = {
    "A-B (Hotspot/Big Island)": {"start": [199.5, 17.5], "end": [203.5, 21.5]},
    "C-D (Oahu/Maui)": {"start": [197.5, 19.5], "end": [201.5, 23.5]},
    "E-F (Older Seamounts)": {"start": [195.5, 21.5], "end": [199.5, 25.5]},
}

# Calculate the minimum bounding box for all profiles and the hotspot
all_lons = [hotspot_lon]
all_lats = [hotspot_lat]
for coords in profiles.values():
    all_lons.extend([coords["start"][0], coords["end"][0]])
    all_lats.extend([coords["start"][1], coords["end"][1]])

# Add a small buffer to the calculated min/max coordinates
buffer = 0.5
region_lon_min = min(all_lons) - buffer
region_lon_max = max(all_lons) + buffer
region_lat_min = min(all_lats) - buffer
region_lat_max = max(all_lats) + buffer

region = [region_lon_min, region_lon_max, region_lat_min, region_lat_max]

# ==========================================
# 2. 繪製平面圖 (包含地形、火山、熱點、沙灘球與剖面線)
# ==========================================
fig = pygmt.Figure()

# 載入高程資料與繪製底圖
grid = pygmt.datasets.load_earth_relief(resolution="01m", region=region)
fig.grdimage(grid=grid, cmap="etopo1", shading="+a45+nt0.5")
fig.coast(shorelines="1p,black", frame=["WSne", "xaf", "yaf"])

# 標示夏威夷熱點 (Hotspot Location)
fig.plot(x=[hotspot_lon], y=[hotspot_lat], style="a0.8c", fill="red", pen="1p,black")
fig.text(x=hotspot_lon, y=hotspot_lat - 0.3, text="Active Hotspot", font="12p,Helvetica-Bold,red")

# 繪製多條剖面線
colors = ["red", "blue", "green"]
for (name, coords), color in zip(profiles.items(), colors):
    lon1, lat1 = coords["start"]
    lon2, lat2 = coords["end"]
    fig.plot(x=[lon1, lon2], y=[lat1, lat2], pen=f"2p,{color},-")
    fig.text(x=lon1, y=lat1-0.2, text=name.split()[0].split('-')[0], font=f"12p,Helvetica-Bold,{color}") # 標示 A, C, E
    fig.text(x=lon2, y=lat2+0.2, text=name.split()[0].split('-')[1], font=f"12p,Helvetica-Bold,{color}") # 標示 B, D, F

# 加入示範性的震源機制解 (Focal Mechanisms / Beachballs)
# 格式: [lon, lat, depth, strike, dip, rake, magnitude]
# 夏威夷大島側翼常有正斷層或滑移
focal_data = dict(
    longitude=[204.5, 204.0],
    latitude=[19.4, 19.8],
    depth=[8.0, 12.0],
    strike=[45, 120],
    dip=[60, 45],
    rake=[-90, -30], # -90 為正斷層
    magnitude=[5.5, 5.0]
)
fig.meca(
    spec=focal_data,
    scale="0.8c",  # 沙灘球大小
    convention="aki",
    cmap="red",    # 壓縮區顏色
    pen="0.5p,black" # Explicitly set a valid pen style
)

fig.savefig("hawaii_map.png")
print("✅ 已生成夏威夷平面圖: hawaii_map.png")

# ==========================================
# 3. 繪製多條剖面地形與地震比較圖 (迴圈處理)
# ==========================================
fig_sec = pygmt.Figure()

# 建立一個 3 行 1 列的子圖 (Subplots) 來比較三條剖面
with fig_sec.subplot(nrows=3, ncols=1, figsize=("15c", "20c"), sharex="b"):
    for i, ((name, coords), color) in enumerate(zip(profiles.items(), colors)):
        with fig_sec.set_panel(panel=i):
            # 取出剖面線兩端
            lon1, lat1 = coords["start"]
            lon2, lat2 = coords["end"]

            # 使用 pygmt.project 在地形網格上取樣 (grdtrack)
            # 1. 產生沿線座標點
            track = pygmt.project(center=[lon1, lat1], endpoint=[lon2, lat2], generate="2k")
            # 2. 內插地形高度
            topo_profile = pygmt.grdtrack(points=track, grid=grid, newcolname='elevation_m')

            # 設定該剖面的座標軸 (X: 距離 0~600km, Y: 深度 -6000m ~ 4500m)
            fig_sec.basemap(
                region=[0, 600, -6000, 4500],
                projection="X15c/5c",
                frame=[f'WSne+t"{name}"', 'xaf', 'yaf+l"Elevation (m)"']
            )

            # 繪製地形剖面塗色 (海底為藍色，海平面上為灰色)
            fig_sec.plot(
                x=topo_profile.iloc[:, 2], # 距離 (km)
                y=topo_profile.iloc[:, 3], # 高程 (m)
                close="+y0",  # 封閉多邊形至 y=0
                pen=f"1p,{color}",
                fill="gray"
            )

            # 畫一條海平面基準線
            fig_sec.plot(x=[0, 600], y=[0, 0], pen="0.5p,black,--")

fig_sec.savefig("hawaii_profiles.png")
print("✅ 已生成剖面對比圖: hawaii_profiles.png")
```
